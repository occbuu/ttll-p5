# Reproduce the Paper5 analysis
This code-only notebook requires authorized local survey inputs and separately obtained public spatial data. It does not download or expose restricted participant data. Copy `config.example.json` to `config.local.json`, edit the paths, and never commit the local file.

```python
from pathlib import Path
import json, subprocess, sys
ROOT = Path.cwd()
CONFIG = ROOT / 'config.local.json'
if not CONFIG.exists():
    raise FileNotFoundError('Create config.local.json from config.example.json and supply authorized paths.')
cfg = json.loads(CONFIG.read_text(encoding='utf-8'))
cfg
```

```python
cmd = [sys.executable, str(ROOT/'analyse_paper5.py'),
       '--clean-authorities', cfg['clean_authorities'],
       '--raw-authorities', cfg['raw_authorities'],
       '--legacy-tables', cfg['legacy_tables'],
       '--ward-geojson', cfg['ward_geojson'],
       '--vietnam-geojson', cfg['vietnam_geojson'],
       '--poi-gpkg', cfg['poi_gpkg'],
       '--output-dir', cfg.get('output_dir','outputs')]
subprocess.run(cmd, check=True)
```

```python
results = json.loads((ROOT/cfg.get('output_dir','outputs')/'results.json').read_text(encoding='utf-8'))
assert results['verified_n'] == 23
assert results['excluded_post_reform_interview_data'] is True
results
```
